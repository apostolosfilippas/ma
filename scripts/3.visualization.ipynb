# Visualising marketplace data

## Why visualize data?

A table or an excel sheet gives us the numbers. But people are (typically)visual learners. A visualization helps us see what is happening: a pattern, a group, a change, or something surprising that deserves a closer look.

This is useful whether you're a data scientist, a product manager, a founder, or simply trying to explain an idea to someone. A good visualization helps you both explore a question more thoroughly and share what you find.

Today we'll work in two parts:

- **Part 1:** Make a tiny dataset and use a scatterplot to see how visualizations work.
- **Part 2:** Use real Airbnb listings to explore prices, choice, and competition.

You don't need to memorize every command. You can ask your AI agent for help with the code; your job is to understand what each dot, bar, or line represents and whether the result makes sense.


# Part 1: Introduction to visualizations
---


## 1. Data

Imagine twelve second-hand phones listed on a marketplace. Each row below is one listing, with the phone's age, asking price, and condition.

```python
# These tools help us make a table and turn it into a visualization.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Each position in these lists describes the same phone.
# All the data are invented; these are asking prices, not completed sales.
demo_listings = pd.DataFrame({
    "age_years": [1, 4, 2, 6, 3, 5, 2, 5, 1, 4, 6, 3],
    "asking_price": [520, 300, 460, 180, 900, 240, 590, 350, 640, 440, 300, 500],
    "condition": ["Good"] * 6 + ["Like new"] * 6,
})

demo_listings
```

## 2. A simple visualization

A **scatterplot** represents each row with a dot. Its position shows some attributes of the listing - in this case, two numerical values:

- **Left to right (the x-axis):** the phone's age in years.
- **Bottom to top (the y-axis):** its asking price in US dollars.

For example, the first row becomes a dot at **1 year and $520**. Once we place all twelve dots, we can look at their pattern instead of comparing the rows one at a time.

We'll use two tools together: **Matplotlib** creates the figure and lets us adjust it; **Seaborn** draws the dots using columns from our DataFrame. In the code, `fig` is the whole figure, and `ax` is the plotting area inside it.


```python
# Start with a clean background and a figure 8 inches wide by 5 inches tall.
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=demo_listings, # the data we are working with
    x="age_years",  # the column that determines the x-axis position of each dot
    y="asking_price", # the column that determines the y-axis position of each dot
    color="#25658A", # the color of the dots
    s=80, # the size of the dots.
    ax=ax,  
)

# Give readers a title and axis labels with units.
ax.set(
    title="Phone age and asking price",
    xlabel="Phone age (years)", ylabel="Asking price (USD)",
    xlim=(0, 7), ylim=(0, 1000),
)
fig.tight_layout() # Leave room for the labels.
plt.show()         # Display the visualization in the notebook.
plt.close(fig)     # Close it afterward so it doesn't appear again.
```

### What can we see now?

Most older phones in our little dataset have lower asking prices. But one three-year-old phone is listed for **$900**. It stands out much more clearly in the visualization.

That unusual point is a reason to ask a question, not automatically delete a row. In real data, we'd want to check whether it was a different model, a special offer, or a mistake.

The visualization makes a pattern easier to see; it doesn't explain its cause. And because we invented these numbers, they tell us nothing about actual phone prices.

## 3. Use color and shape to show additional attributes

So far, every dot looks the same. Let's use **color and marker shape** to show the phone's condition. We'll keep the positions and axis limits unchanged, so we're looking at the same data in a new way.

In Seaborn, `hue="condition"` chooses colors using that column, and `style="condition"` chooses marker shapes. The legend explains both. Color now tells us something about the data, rather than just decorating the picture.


```python
# Draw the same phones again, this time showing their condition.
fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=demo_listings, x="age_years", y="asking_price",
    hue="condition", style="condition",
    palette={"Good": "#25658A", "Like new": "#C56B24"},
    markers={"Good": "o", "Like new": "s"},  # Circles and squares.
    s=90, ax=ax,
)

# Keep the same scales so changes in the design don't change the comparison.
ax.set(
    title="Phone age and asking price, by condition\n12 invented listings — practice data",
    xlabel="Phone age (years)", ylabel="Asking price (USD)",
    xlim=(0, 7), ylim=(0, 1000),
)
ax.legend(title="Condition", frameon=False)
fig.tight_layout()
plt.show()
plt.close(fig)
```

# Part 2: Exploring Airbnb's marketplace

We'll use real listings to explore prices, choice, and competition. Different questions call for different visualizations.

### Data

As in the DataFrames lesson, the data come from [Inside Airbnb](https://insideairbnb.com/get-the-data/). We'll start with NYC listings from **August 10, 2026**, then compare twelve monthly snapshots from **September 2025–August 2026**.

Each row represents a listing, not a booking. The advertised nightly price is not the full trip cost or the host's revenue.


```python
# Import the tools for the Airbnb analysis.
from pathlib import Path

import matplotlib.dates as mdates
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter, StrMethodFormatter
import numpy as np
import pandas as pd
import seaborn as sns

# Locate the course folder from either ma/ or ma/scripts/.
course_folder = Path.cwd()
if not (course_folder / "data").is_dir():
    course_folder = course_folder.parent

data_folder = course_folder / "data"
history_folder = data_folder / "inside-airbnb-nyc-history"
output_folder = course_folder / "temp/lecture-3-viz"
output_folder.mkdir(parents=True, exist_ok=True)

# Keep colors consistent across visualizations.
BLUE = "#25658A"
ORANGE = "#C56B24"
INK = "#27343D"
MISSING = "#F7DEDE"

# Set the font, grid, and axis styles.
sns.set_theme(style="whitegrid", context="notebook", font="DejaVu Sans")
plt.rcParams.update({
    "axes.spines.top": False,
    "axes.spines.right": False,
    "grid.alpha": 0.25,
    "text.color": INK,
    "text.parse_math": False,  # Show dollar signs as currency, not math markup.
    "axes.labelcolor": INK,
    "axes.titleweight": "bold",
    "figure.dpi": 110,
})
```

```python
# Reuse this function to format, save, and display the Airbnb visualizations.
def show_visualization(fig, filename, title, subtitle):
    """Add a title, subtitle, and source; save and display the figure.

    Args:
        fig: A Matplotlib figure, usually created with plt.subplots().
        filename: Output filename without the .png or .pdf extension.
        title: Main heading describing the visualization.
        subtitle: Context such as dates, sample, units, or price range.

    Saves PNG and PDF files in output_folder, displays the figure,
    and closes it to prevent duplicate displays.
    """
    # Keep a fixed gap between the title and subtitle at any figure height.
    figure_height_points = fig.get_figheight() * 72
    subtitle_y = 0.98 - 28 / figure_height_points
    fig.suptitle(title, x=0.08, y=0.98, ha="left", va="top", fontsize=16, weight="bold")
    fig.text(0.08, subtitle_y, subtitle, ha="left", va="top", fontsize=10)
    fig.text(0.08, 0.015, "Source: Inside Airbnb | NYC | CC BY 4.0", fontsize=9)
    fig.tight_layout(rect=[0, 0.05, 1, subtitle_y])
    fig.savefig(output_folder / f"{filename}.png", dpi=160, bbox_inches="tight")
    fig.savefig(output_folder / f"{filename}.pdf", bbox_inches="tight")
    plt.show()
    plt.close(fig)
```

```python
# Load the snapshot used in the DataFrames lesson.
listings = pd.read_csv(data_folder / "inside-airbnb-nyc-listings-2026-08-10.csv")
listings = listings.rename(columns={
    "id": "listing_id",
    "neighbourhood_group": "borough",
    "neighbourhood": "neighborhood",
    "price": "nightly_price",
    "number_of_reviews_ltm": "reviews_last_year",
})

# Count all listings when measuring supply.
# For price comparisons, keep recorded prices from $20 to $1,000.
# This makes the central part of the distribution easier to see.
# Prices outside this range are not necessarily errors.
analysis_listings = listings[listings["nightly_price"].between(20, 1000)].copy()

print(f"All captured listings: {len(listings):,}")
print(f"Price-analysis sample: {len(analysis_listings):,}")
print(f"Missing prices: {listings['nightly_price'].isna().mean():.1%}")

analysis_listings[["borough", "neighborhood", "room_type", "nightly_price"]].head()
```

## 1. Comparing counts using bar plots

A **bar plot** compares categories using bar lengths. Here, each bar represents a borough, and its length shows the number of listings.

Horizontal bars leave room for the borough names. Sorting them makes the ranking clear. The count axis starts at zero because readers compare the lengths of the bars.


```python
# Count all listings, including those without a recorded price.
borough_supply = listings["borough"].value_counts()
borough_order = borough_supply.index.tolist()

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.barh(borough_supply.index, borough_supply.to_numpy(), color=BLUE)
ax.invert_yaxis()  # Put the largest category at the top.
ax.bar_label(bars, labels=[f"{n:,}" for n in borough_supply.values], padding=5)
ax.set_xlim(0, borough_supply.max() * 1.16)  # Leave room for the labels.
ax.set_xlabel("Captured listings")
ax.xaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
ax.grid(axis="y", visible=False)

show_visualization(fig, "01-supply", "Airbnb listings by borough",
           f"August 10, 2026 | All {len(listings):,} captured listings")
```

Listing counts measure advertised supply. They don't tell us how many places are available for particular travel dates or how many bookings each borough receives.

## 2. Visualizing distributions with histograms

An average summarizes prices with one number. A **histogram** shows how prices are distributed: it divides them into ranges, called *bins*, and counts the listings in each range.

Unlike a bar plot, which compares categories, a histogram uses a numerical scale. Here, the x-axis shows price and the y-axis shows the number of listings. We'll use $50 bins to see which prices are common and how far the distribution extends.

```python
# Change the bin width to 25 or 100 to compare the distributions.
bin_width = 50

mean_price = analysis_listings["nightly_price"].mean()
median_price = analysis_listings["nightly_price"].median()

fig, ax = plt.subplots(figsize=(10, 5))
sns.histplot(data=analysis_listings, x="nightly_price",
             bins=np.arange(0, 1000 + bin_width, bin_width), color=BLUE, ax=ax)
ax.axvline(mean_price, color=ORANGE, linestyle="--",
           label=f"Mean: {mean_price:.0f} USD")
ax.axvline(median_price, color=INK, linestyle=":",
           label=f"Median: {median_price:.0f} USD")
ax.set(xlim=(0, 1000), xlabel="Advertised nightly price (USD)", ylabel="Listings")
ax.legend(frameon=False)

show_visualization(fig, "02-price-distribution", "One average does not describe every listing",
           f"August 10, 2026 | Recorded prices $20–$1,000 | n = {len(analysis_listings):,}")
```

## 3. Making comparisons using panels within a plot

We can squeeze in several plots within one using panels. Here, the left panel shows the number of listings within budget; the right shows the percentage. The borough order stays the same but the units differ.

A borough can have the most affordable options without having the highest percentage of affordable options. We'll compare private rooms, using each borough's private-room listings with recorded prices from $20 to $1,000 as the denominator.


```python
# Set the budget and room type for the comparison.
budget = 150
chosen_room_type = "Private room"

room_listings = analysis_listings[
    analysis_listings["room_type"] == chosen_room_type
].copy()
room_listings["within_budget"] = room_listings["nightly_price"] <= budget

affordability = room_listings.groupby("borough").agg(
    priced_listings=("listing_id", "size"),
    affordable_listings=("within_budget", "sum"),
    affordable_share=("within_budget", "mean"),
).sort_values("affordable_listings", ascending=False)

# Express the within-budget share as a percentage.
affordability["affordable_pct"] = affordability["affordable_share"] * 100
affordability[["priced_listings", "affordable_listings", "affordable_pct"]].round(1)
```

```python
# Use the same borough order in both panels.
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5), sharey=True)

bars = axes[0].barh(affordability.index, affordability["affordable_listings"], color=BLUE)
axes[0].bar_label(bars, labels=[f"{n:,}" for n in affordability["affordable_listings"]], padding=5)
axes[0].set_xlim(0, affordability["affordable_listings"].max() * 1.2)
axes[0].set_xlabel("Number within budget")
axes[0].xaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
axes[0].invert_yaxis()

bars = axes[1].barh(affordability.index, affordability["affordable_pct"], color=BLUE)
axes[1].bar_label(bars, labels=[f"{p:.1f}%" for p in affordability["affordable_pct"]], padding=5)
axes[1].set(xlim=(0, 110), xlabel="Percentage within budget")
axes[1].xaxis.set_major_formatter(PercentFormatter(100))
for ax in axes:
    ax.grid(axis="y", visible=False)

show_visualization(fig, "03-affordability", f"Advertised options at {budget} USD or less",
           f"August 10, 2026 | {chosen_room_type} | Denominator: each borough's prices $20–$1,000")
```

## 4. Measuring market concentration

A **concentration curve** shows how much of a total belongs to a given share of participants. Here, host IDs are ranked from most to fewest listings.

- The **x-axis** shows the cumulative percentage of host IDs, starting with the largest.
- The **y-axis** shows their combined percentage of listings.

How many hosts account for half the listings? Start at 50% on the y-axis, move across to the curve, then down to the x-axis. The code finds the smallest number of hosts needed to reach at least half the listings.

The diagonal represents equal numbers of listings per host. The farther the curve lies above it, the more concentrated supply is.


```python
# Count all listings per host ID, including those with missing prices.
# groupby excludes missing host IDs; this snapshot has none.
host_sizes = listings.groupby("host_id").size().sort_values(ascending=False)
host_percent = np.arange(1, len(host_sizes) + 1) / len(host_sizes) * 100
cumulative_listings = host_sizes.cumsum()
listing_percent = cumulative_listings / host_sizes.sum() * 100

# Find the first whole number of hosts whose listings reach at least half the total.
hosts_for_half = int((cumulative_listings < host_sizes.sum() / 2).sum()) + 1
half_host_pct = hosts_for_half / len(host_sizes) * 100
half_listing_count = cumulative_listings.iloc[hosts_for_half - 1]
half_listing_pct = half_listing_count / host_sizes.sum() * 100

print(f"Largest {hosts_for_half:,} of {len(host_sizes):,} hosts ({half_host_pct:.1f}%)")
print(f"Their listings: {half_listing_count:,} of {host_sizes.sum():,} ({half_listing_pct:.1f}%)")

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.plot(np.r_[0, host_percent], np.r_[0, listing_percent], color=BLUE, linewidth=2.5)
ax.plot([0, 100], [0, 100], color=INK, linestyle="--", label="Equal listings per host ID")
ax.scatter(half_host_pct, half_listing_pct, color=ORANGE, zorder=3)
ax.annotate(f"Largest {half_host_pct:.1f}% of hosts\nown {half_listing_pct:.0f}% of listings",
            xy=(half_host_pct, half_listing_pct), xytext=(35, 18),
            arrowprops={"arrowstyle": "->", "color": INK}, fontsize=11)
ax.set(xlim=(0, 100), ylim=(0, 100),
       xlabel="Host IDs, ranked from most to fewest listings (cumulative %)",
       ylabel="Captured listings (cumulative %)")
ax.xaxis.set_major_formatter(PercentFormatter(100))
ax.yaxis.set_major_formatter(PercentFormatter(100))
ax.legend(loc="lower right", frameon=False)

show_visualization(fig, "06-host-concentration", "How concentrated is Airbnb supply?",
           f"August 10, 2026 | {len(host_sizes):,} host IDs | {host_sizes.sum():,} listings")
```

Here, "own" means the listings are associated with a host account, not verified property ownership. A manager may represent several owners, and a business may use several accounts.

This measures concentration in **listings**, not bookings or revenue. What additional data would be needed to study those?


## 5. Plotting quantities over time

A **line plot** connects values along an ordered axis. Here, the x-axis shows time, and each marker represents a snapshot collected between September 2025 and August 2026. The lines connect those observations; they are not daily measurements.

The left panel shows each room type's percentage of all listings. The right shows listing counts. Colors and line styles identify the same room types in both panels.

Pandas reads the compressed `.csv.gz` files directly. Input files stay in `data/`; results are saved in `temp/lecture-3-viz/`.

```python
# Use the actual collection dates, which are not evenly spaced.
snapshot_dates = [
    "2025-09-01", "2025-10-01", "2025-11-01", "2025-12-04",
    "2026-01-15", "2026-02-13", "2026-03-16", "2026-04-14",
    "2026-05-14", "2026-06-14", "2026-07-12", "2026-08-10",
]

# Store each snapshot in a dictionary keyed by its collection date.
snapshots = {}
for date in snapshot_dates:
    source_file = history_folder / f"listings-{date}.csv.gz"
    if date == "2026-08-10":
        # Reuse the August input from the DataFrames lesson.
        source_file = data_folder / "inside-airbnb-nyc-listings-2026-08-10.csv"

    snapshot = pd.read_csv(source_file, usecols=[
        "id", "host_id", "neighbourhood_group", "room_type", "price",
    ]).rename(columns={"neighbourhood_group": "borough"})
    snapshots[pd.Timestamp(date)] = snapshot

print(f"Loaded {len(snapshots)} original snapshots.")
```

```python
# Count listings by borough and room type in each snapshot.
# Retain borough/room-type combinations with zero listings.
boroughs = ["Bronx", "Brooklyn", "Manhattan", "Queens", "Staten Island"]
room_types = ["Entire home/apt", "Private room", "Shared room", "Hotel room"]
pairs = pd.MultiIndex.from_product([boroughs, room_types], names=["borough", "room_type"])

count_tables = []
for date, snapshot in snapshots.items():
    counts = snapshot.groupby(["borough", "room_type"]).size()
    counts = counts.reindex(pairs, fill_value=0).reset_index(name="listing_count")
    counts["snapshot_date"] = date
    count_tables.append(counts)

history = pd.concat(count_tables, ignore_index=True)
history.to_csv(output_folder / "marketplace-counts.csv", index=False)

# Sum across boroughs to get NYC totals for each room type.
room_history = history.groupby(["snapshot_date", "room_type"], as_index=False)[
    "listing_count"
].sum()

# Calculate shares using all four room types, although only two are plotted.
snapshot_totals = room_history.groupby("snapshot_date")["listing_count"].transform("sum")
room_history["share_pct"] = room_history["listing_count"] / snapshot_totals * 100

# Keep the same color and line style for each room type.
room_styles = [
    ("Entire home/apt", BLUE, "o", "-"),
    ("Private room", ORANGE, "s", "--"),
]
```

```python
# Compare shares and counts in separate panels.
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
for room_type, color, marker, line_style in room_styles:
    series = room_history[room_history["room_type"] == room_type].sort_values("snapshot_date")
    axes[0].plot(series["snapshot_date"], series["share_pct"],
                 color=color, marker=marker, linestyle=line_style, label=room_type)
    axes[1].plot(series["snapshot_date"], series["listing_count"],
                 color=color, marker=marker, linestyle=line_style, label=room_type)

axes[0].set(ylabel="Share of all captured listings", ylim=(0, 100))
axes[0].yaxis.set_major_formatter(PercentFormatter(100))
axes[1].set(ylabel="Captured listings", ylim=(0, None))
axes[1].yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

# Mark the May drop without assuming its cause.
may_date = pd.Timestamp("2026-05-14")
may_entire = room_history.loc[
    (room_history["snapshot_date"] == may_date)
    & (room_history["room_type"] == "Entire home/apt"), "listing_count"
].iloc[0]
axes[1].annotate("May drop:\ninvestigate coverage",
                 xy=(may_date, may_entire), xytext=(0.25, 0.34),
                 textcoords="axes fraction",
                 arrowprops={"arrowstyle": "->", "color": INK}, fontsize=10)

for ax in axes:
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b\n%Y"))
    ax.legend(loc="lower left", frameon=False, fontsize=10)

show_visualization(fig, "07-counts-and-shares", "Listings over time",
           "September 2025–August 2026 | All captured listings | Markers show actual snapshot dates")
```

```python
# Compare the first and last snapshots to check the title's claim.
entire_history = room_history[room_history["room_type"] == "Entire home/apt"]
entire_history.iloc[[0, -1]][["snapshot_date", "listing_count", "share_pct"]].round(1)
```

Entire apartments fall from **19,193 to 16,693 listings**, while their share rises from **53.1% to 55.2%**. Other categories shrank more, so entire apartments account for a larger share of a smaller marketplace.

The sharp May drop could reflect changes in the marketplace, data collection, or both. These files cannot distinguish between those explanations. One year of observations is also insufficient to establish a repeating seasonal pattern.

## 6. Missing data

### Line plot with a missing-data gap

The December, January, and February snapshots contain listings but **no recorded prices**. A missing price is not a zero price.

The price line therefore has a gap, marked by a pale red band. Connecting the line across those months would imply information we don't have.

The lower panel shows the percentage of listings with a recorded price. Both panels use the same dates so we can compare price movements with changes in data coverage.


```python
# Calculate the NYC median from individual listing prices.
# An average of borough medians would measure something different.
monthly_rows = []

for date, snapshot in snapshots.items():
    recorded = snapshot["price"].notna()
    usable = snapshot["price"].between(20, 1000)

    monthly_rows.append({
        "snapshot_date": date,
        # With no usable prices, the median is NaN (a missing value).
        "median_price": snapshot.loc[usable, "price"].median(),
        "price_sample_count": int(usable.sum()),
        # Coverage is the percentage of all listings with any recorded price.
        "recorded_price_pct": recorded.mean() * 100,
    })

price_history = pd.DataFrame(monthly_rows).sort_values("snapshot_date")
missing_months = price_history.loc[
    price_history["median_price"].isna(), "snapshot_date"
].dt.to_period("M")

# Save derived results in temp/, leaving input files unchanged.
price_history.to_csv(output_folder / "price-history.csv", index=False)
price_history.round(1)
```

```python
# Matplotlib breaks the line at NaN values.
# Keep missing values rather than dropping the rows or filling the gaps.
fig, axes = plt.subplots(2, 1, figsize=(11, 7.5), sharex=True,
                         gridspec_kw={"height_ratios": [2, 1]})

axes[0].plot(price_history["snapshot_date"], price_history["median_price"],
             color=BLUE, marker="o", linewidth=2)
axes[0].set(ylabel="Median nightly price (USD)", ylim=(0, 300))

# Shade the months with no recorded prices.
for month in missing_months:
    for ax in axes:
        ax.axvspan(month.start_time, (month + 1).start_time, color=MISSING, zorder=0)

gap_center = pd.Timestamp("2026-01-15")
axes[0].text(gap_center, 0.83, "← Price data →\nunavailable",
             transform=axes[0].get_xaxis_transform(), ha="center",
             color="#8E3B46", fontsize=11)

# Zero coverage means no listings have a recorded price.
axes[1].plot(price_history["snapshot_date"], price_history["recorded_price_pct"],
             color=INK, marker="s", linestyle="--")
axes[1].set(ylabel="Listings with\na recorded price", ylim=(0, 100))
axes[1].yaxis.set_major_formatter(PercentFormatter(100))
axes[1].xaxis.set_major_locator(mdates.MonthLocator())
axes[1].xaxis.set_major_formatter(mdates.DateFormatter("%b\n%Y"))

show_visualization(fig, "08-prices-with-missing-data", "A missing price is not a zero price",
           "September 2025–August 2026 | Median: prices $20–$1,000 | Coverage: all captured listings")
```

Even outside the shaded months, many listings have no recorded price. A change in the median may reflect a different mix of observed listings, rather than the same hosts changing their prices.

The red band marks missing price data, not an absence of listings or bookings.

## 7. Improving the design

### Bar plots: before and after

Both versions show the same borough counts. The first uses vertical bars in alphabetical order. The second uses horizontal bars sorted by size, with values labeled directly.

Orientation, order, labels, and color affect how easily readers can compare the data. Which choices make the largest category easier to identify? Do the colors in the first version add information?


```python
# Use the same numbers in both versions, with count axes starting at zero.
alphabetical_supply = borough_supply.sort_index()
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Before: alphabetical order, tilted labels, and a different color per bar.
# The colors add no information beyond the borough labels.
axes[0].bar(alphabetical_supply.index, alphabetical_supply.values,
            color=["#B47AA6", "#63A8A3", "#D3A04A", "#8F91BD", "#BF7C66"])
axes[0].tick_params(axis="x", labelrotation=45)
axes[0].set(title="Before", ylabel="Captured listings")
axes[0].yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

# After: sorted horizontal bars with direct value labels.
bars = axes[1].barh(borough_supply.index, borough_supply.values, color=BLUE)
axes[1].invert_yaxis()
axes[1].bar_label(bars, labels=[f"{n:,}" for n in borough_supply.values], padding=5)
axes[1].set(title="After", xlabel="Captured listings",
            xlim=(0, borough_supply.max() * 1.2))
axes[1].xaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
axes[1].set_xticks([0, 5000, 10000, 15000])  # Avoid crowded tick labels.
axes[1].grid(axis="y", visible=False)

show_visualization(fig, "09-visualization-makeover", "Same data, less work for the reader",
           f"August 10, 2026 | All {len(listings):,} captured listings | Both count axes start at zero")
```
